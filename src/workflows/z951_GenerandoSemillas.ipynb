# Generando 100 semillas

A partir de la semilla primigenia **300089** (la del notebook senior z950)
<br>Se generan 100 semillas y se guardan en el archivo `semillas`

```r
# instalo y cargo la libreria  primes
if (!require("primes")) install.packages("primes")
require("primes")
```

```r
# genero numeros primos
primos <- generate_primes(min = 100000, max = 1000000)

set.seed(300089) # semilla primigenia del notebook senior

# me quedo con 100 primos al azar
semillas <- sample(primos, 100)

print(semillas)
```

```r
# guardo las semillas en el archivo  semillas
writeLines(as.character(semillas), "semillas")
```

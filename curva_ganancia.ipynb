```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("datasets/kaggle/submissions.csv")

```

```python
# filter only "KA"

df_ka = df[df.fileName.str.contains("KA")].copy()

df_ka["publicScore"] = pd.to_numeric(df_ka["publicScore"])

df_ka["experimento"] = df['fileName'].str.split('_').str[0]

df_ka['envios'] = pd.to_numeric(df_ka['fileName'].str.split('_').str[1].str.replace('.csv', ''))

df_ka = df_ka[df_ka['experimento'].str.contains("KA")]

df_ka = df_ka[df_ka.envios.isin([12000, 11500, 11000, 10500, 10000, 9500, 9000])]

df_ka
```

```python
import seaborn as sns

sns.lineplot(data=df_ka[df_ka["experimento"] == "KA5944"].sort_values(by='envios'), x='envios', y='publicScore', hue='experimento')
```
